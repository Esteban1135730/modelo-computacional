# Tarea: Problema de Minimización de Presupuesto

## Enunciado del problema

Una empresa necesita adquirir materiales para un proyecto. Debe comprar dos tipos de insumos (A y B) minimizando el costo total, cumpliendo con requisitos mínimos de abastecimiento.

---

## Formulación matemática

### Variables de decisión

$$
x_1 = \text{cantidad de unidades del material A (en kg)}
$$

$$
x_2 = \text{cantidad de unidades del material B (en kg)}
$$

### Función objetivo (minimizar)

$$
\min \quad Z = 0.5\,x_1 + 0.5\,x_2
$$

Donde los coeficientes 0.5 representan el costo unitario (en unidades monetarias) de cada material.

### Restricciones

**Restricción 1:** Se requieren al menos 2 unidades en total para el proyecto.

$$
x_1 + x_2 \geq 2
$$

**Restricción 2:** Debe comprarse al menos 1 unidad del material A (por especificación técnica).

$$
x_1 \geq 1
$$

**Restricción 3:** Debe comprarse al menos 1 unidad del material B (por especificación técnica).

$$
x_2 \geq 1
$$

### Naturaleza de las variables

$$
x_1 \geq 0 \quad \text{y} \quad x_2 \geq 0
$$

Las variables son **continuas y no negativas** (no tiene sentido comprar cantidades negativas de materiales).

### Modelo completo

$$
\boxed{
\begin{aligned}
\min \quad & Z = 0.5\,x_1 + 0.5\,x_2 \\
\text{s.a.} \quad & x_1 + x_2 \geq 2 \\
& x_1 \geq 1 \\
& x_2 \geq 1 \\
& x_1, \, x_2 \geq 0
\end{aligned}
}
$$

---

## Solución analítica

Las restricciones $x_1 \geq 1$ y $x_2 \geq 1$ implican que $x_1 + x_2 \geq 2$, por lo que la restricción 1 se cumple automáticamente cuando se satisfacen las restricciones 2 y 3.

El mínimo de $Z = 0.5x_1 + 0.5x_2$ sujeto a $x_1 \geq 1$ y $x_2 \geq 1$ se alcanza cuando $x_1$ y $x_2$ son lo más pequeños posible:

$$
x_1^* = 1, \quad x_2^* = 1
$$

**Valor óptimo:**

$$
Z^* = 0.5(1) + 0.5(1) = \boxed{1}
$$

---

## Verificación computacional con Python

```python
from scipy.optimize import linprog

# Coeficientes de la función objetivo: min 0.5*x1 + 0.5*x2
c = [0.5, 0.5]

# Restricciones en forma A_ub @ x <= b_ub
# Convertimos >= a <= multiplicando por -1:
# -x1 - x2 <= -2  =>  x1 + x2 >= 2
# -x1 <= -1       =>  x1 >= 1
# -x2 <= -1       =>  x2 >= 1
A_ub = [[-1, -1], [-1, 0], [0, -1]]
b_ub = [-2, -1, -1]

# Límites de variables (por defecto x >= 0)
bounds = [(0, None), (0, None)]

resultado = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method='highs')

print("Estado:", resultado.message)
print("Solución óptima: x1 =", resultado.x[0], ", x2 =", resultado.x[1])
print("Valor óptimo Z* =", resultado.fun)
```
